````python
# largest local sum
def lls(nums: list[float]) -> tuple[float, int, int]:
    best_start = best_end = 0
    best_sum = nums[best_start]

    for start in range(0, len(nums)):
        current_sum = nums[start]
        for end in range(start + 1, len(nums)):
            current_sum += nums[end] # current_sum = current sum + nums[end]
            if current_sum > best_sum:
                best_sum = current_sum
                best_start = start
                best_end = end

    return best_sum, best_start, best_end

print(lls([-2, 1, -3, 4, -1, 2, 1, -5, 4])) # (6, 3, 6)

def test1():
    # Testcase regular case
    print("Test 1: ", lls([-2, 1, -3, 4, -1, 2, 1, -5, 4]) == (6, 3, 6))

def test2():
    # Test edgecase with one element
    print("Test 2: ", lls([5]) == (5, 0, 0))

test1()
test2()

````

````python
# largest local sum with one loop only
def lls2(nums: list[float]) -> tuple[float, int, int]:
    # check if nums is not empty otherwise throw an error
    if not nums:
        raise ValueError("Input list cannot be empty")

    best_start = best_end = current_start = 0
    best_sum = current_sum = nums[best_start]

    for i in range(1, len(nums)):
        if current_sum < 0:
            current_sum = nums[i]
            current_start = i
        else:
            current_sum += nums[i]

        if current_sum > best_sum:
            best_sum = current_sum
            best_start = current_start
            best_end = i

    return best_sum, best_start, best_end

def test3():
    # Testcase regular case
    print("Test 1: ", lls2([-2, 1, -3, 4, -1, 2, 1, -5, 4]) == (6, 3, 6))

def test4():
    # Test edgecase with one element
    print("Test 2: ", lls2([5]) == (5, 0, 0))

test3()
test4()
````

````python
def gcd(m, n, depth):
    if n > m:
        print("m > n branch at depth", depth)
        return gcd(n, m, depth + 1)
    elif m % n == 0:
        print ("m % n branch at depth", depth)
        return n
    else:
        print("else branch at depth", depth)
        return gcd(n, m % n, depth + 1)

print(gcd(143, 9555, 0))
````

````python

def largest_local_sum(numbers):
    current_sum = numbers [0]
    max_sum = numbers [0]

    start = 0
    max_start = 0
    max_end = 0

    for i in range (1, len(numbers)):
        if current_sum + numbers [i] < numbers [i]:
            current_sum = numbers [i]
            start = i
        else:
            current_sum += numbers [i]

        if current_sum > max_sum:
            max_sum = current_sum
            max_start = start
            max_end = i

    return max_sum, max_start, max_end

numbers = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

print(largest_local_sum(numbers))
````

### Problem 10

``` {.bash language="bash"}
 * *
```

### Problem 14
Write code that builds the complement sequence for a given DNA sequence. Print the result as follows:

``` {.bash language="bash"}
5'-ATCGTGAATCG-3'
```

### Problem 16
Checking for sequence similarity for short DNA/RNA sequences is oftentimes done using a Dot-Matrix. Given two sequences A and B implement a comparison algorithm that prints the sequence A as the first line in the output and the sequence B down on the left side. Fill the matrix whenever the A\[i\]==B\[j\] with the symbol . The output for the two example sequences A=\"GATTCGCTTAGT\" and B=\"CTGATTCCTTAGTCAG\" should look like this:

``` {.awk language="Awk"}
 GATTCGCTTAGT
```

````python
A = "GATTCGCTTAGT"
B = "CTGATTCCTTAGTCAG"

print("  " + A)

for b in B:
    print(b, end=" ")

    for a in A:
        if b == a:
            print("*", end="")
        else:
            print(" ", end="")

    print()
````
